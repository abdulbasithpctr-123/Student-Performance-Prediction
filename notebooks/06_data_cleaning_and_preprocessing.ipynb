```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Visualization settings

sns.set_style("whitegrid")

# Load original dataset

df = pd.read_csv("../dataset/StudentPerformanceFactors.csv")

print("Original dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
# Create a separate working copy

df_clean = df.copy()

print("Working copy created successfully!")

print("Original dataset shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

```python
# Check dataset information

df_clean.info()

# Check missing values

print("\nMissing Values:")
print(df_clean.isnull().sum())

# Check duplicate records

print("\nDuplicate Records:")
print(df_clean.duplicated().sum())
```

```python
# Standardize categorical values

categorical_columns = df_clean.select_dtypes(
    include=["object", "str"]
).columns.tolist()

for column in categorical_columns:

    df_clean[column] = df_clean[column].str.strip()

    df_clean[column] = df_clean[column].replace(
        "", np.nan
    )

print("Categorical values standardized successfully!")
```

```python
# Check categorical values after standardization

for column in categorical_columns:

    print("\nColumn:", column)
    print(df_clean[column].value_counts(dropna=False))
```

```python
# Numerical feature validation

numerical_columns = df_clean.select_dtypes(
    include=["number"]
).columns.tolist()

for column in numerical_columns:

    print("\nFeature:", column)
    print("Minimum:", df_clean[column].min())
    print("Maximum:", df_clean[column].max())
    print("Unique values:", df_clean[column].nunique())
    print("Missing values:", df_clean[column].isnull().sum())
```

```python
# Check numerical feature constraints

print("Hours studied below 0:",
      (df_clean["Hours_Studied"] < 0).sum())

print("Attendance below 0:",
      (df_clean["Attendance"] < 0).sum())

print("Attendance above 100:",
      (df_clean["Attendance"] > 100).sum())

print("Sleep hours below 0:",
      (df_clean["Sleep_Hours"] < 0).sum())

print("Previous scores below 0:",
      (df_clean["Previous_Scores"] < 0).sum())

print("Previous scores above 100:",
      (df_clean["Previous_Scores"] > 100).sum())

print("Exam scores above 100:",
      (df_clean["Exam_Score"] > 100).sum())
```

```python
# Separate input features and target variable

X = df_clean.drop(columns=["Exam_Score"])

y = df_clean["Exam_Score"]

print("Input features shape:", X.shape)
print("Target variable shape:", y.shape)

print("\nTarget variable:")
print(y.name)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
```

```python
from sklearn.impute import SimpleImputer

# Identify categorical columns

categorical_columns = X_train.select_dtypes(
    include=["object", "str"]
).columns.tolist()

# Create mode imputer

categorical_imputer = SimpleImputer(
    strategy="most_frequent"
)

# Fit only on training data

X_train[categorical_columns] = categorical_imputer.fit_transform(
    X_train[categorical_columns]
)

# Apply training imputation to test data

X_test[categorical_columns] = categorical_imputer.transform(
    X_test[categorical_columns]
)

print("Categorical missing values imputed successfully!")
```

```python
print("Training missing values:")
print(X_train.isnull().sum().sum())

print("\nTesting missing values:")
print(X_test.isnull().sum().sum())
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

# Identify numerical and categorical features

numerical_columns = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_columns = X_train.select_dtypes(
    include=["object", "str"]
).columns.tolist()

print("Numerical features:", numerical_columns)
print("\nCategorical features:", categorical_columns)
```

```python
# Create preprocessing transformer

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_columns
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_columns
        )
    ]
)

# Fit on training data and transform

X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data

X_test_processed = preprocessor.transform(X_test)

print("Training data shape:", X_train_processed.shape)
print("Testing data shape:", X_test_processed.shape)
```

```python
# Get transformed feature names

feature_names = preprocessor.get_feature_names_out()

print("Total processed features:", len(feature_names))

print("\nFirst 20 processed feature names:")
print(feature_names[:20])

print("\nTraining data shape:", X_train_processed.shape)
print("Testing data shape:", X_test_processed.shape)
```

```python
# Check numerical feature scaling

scaled_numerical_data = X_train_processed[
    :, :len(numerical_columns)
]

print("Scaled numerical feature means:")
print(np.mean(scaled_numerical_data, axis=0).round(3))

print("\nScaled numerical feature standard deviations:")
print(np.std(scaled_numerical_data, axis=0).round(3))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

# Numerical preprocessing

numerical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

# Categorical preprocessing

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)

# Combine preprocessing steps

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer, numerical_columns),
        ("cat", categorical_transformer, categorical_columns)
    ]
)

print("Reusable preprocessing pipeline created!")
```

```python
# Fit preprocessing only on training data

X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data

X_test_processed = preprocessor.transform(X_test)

print("Training shape:", X_train_processed.shape)
print("Testing shape:", X_test_processed.shape)
print("Preprocessing completed successfully!")
```

```python
import joblib
import os

os.makedirs("../models", exist_ok=True)

joblib.dump(
    preprocessor,
    "../models/preprocessor.pkl"
)

print("Preprocessing pipeline saved successfully!")
```
