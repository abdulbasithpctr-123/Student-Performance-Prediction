```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("C:\\Users\\pro\\Student-Performance-Prediction\\dataset\\student_performance_prediction.csv")

print("Original Dataset Shape:", df.shape)

df.head()
```

```python
# Keep the original data unchanged
df_clean = df.copy()

print("Working copy created successfully!")
```

```python
# Count missing values
missing_values = df_clean.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)
```

```python
missing_percentage = (
    df_clean.isnull().sum() / len(df_clean)
) * 100

missing_report = pd.DataFrame({
    "Missing Values": df_clean.isnull().sum(),
    "Missing Percentage": missing_percentage
})

display(missing_report)
```

```python
numerical_columns = [
    "Study Hours per Week",
    "Attendance Rate",
    "Previous Grades"
]

for col in numerical_columns:
    print(f"\nColumn: {col}")
    print("Minimum:", df_clean[col].min())
    print("Maximum:", df_clean[col].max())
    print("Negative values:", (df_clean[col] < 0).sum())
```

```python
# Study hours cannot be negative or exceed 168 hours per week
df_clean.loc[
    (df_clean["Study Hours per Week"] < 0) |
    (df_clean["Study Hours per Week"] > 168),
    "Study Hours per Week"
] = np.nan

# Attendance must be between 0 and 100
df_clean.loc[
    (df_clean["Attendance Rate"] < 0) |
    (df_clean["Attendance Rate"] > 100),
    "Attendance Rate"
] = np.nan

# Previous grades are assumed to be percentages
df_clean.loc[
    (df_clean["Previous Grades"] < 0) |
    (df_clean["Previous Grades"] > 100),
    "Previous Grades"
] = np.nan

print("Invalid numerical values converted to NaN.")
```

```python
print("Target values before cleaning:")
print(df_clean["Passed"].value_counts(dropna=False))

# Remove rows where target is missing
df_clean = df_clean.dropna(subset=["Passed"])

print("\nTarget values after cleaning:")
print(df_clean["Passed"].value_counts())

print("\nUpdated Dataset Shape:", df_clean.shape)
```

```python
categorical_columns = [
    "Participation in Extracurricular Activities",
    "Parent Education Level",
    "Passed"
]

for col in categorical_columns:
    print(f"\n{col}")
    print(df_clean[col].value_counts(dropna=False))
```

```python
print("Duplicate records:", df_clean.duplicated().sum())
```

```python
print("FINAL DATA CLEANING REPORT")
print("=" * 40)

print("Original Dataset Shape:", df.shape)
print("Cleaned Dataset Shape:", df_clean.shape)

print("\nMissing Values:")
print(df_clean.isnull().sum())

print("\nDuplicate Records:")
print(df_clean.duplicated().sum())

print("\nNumerical Summary:")
display(df_clean.describe())
```

```python
df_clean.to_csv(
    "../dataset/student_performance_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```

```python
print("Total Student IDs:", df_clean["Student ID"].count())

print(
    "Unique Student IDs:",
    df_clean["Student ID"].nunique()
)

print(
    "Duplicate Student IDs:",
    df_clean["Student ID"].duplicated().sum()
)
```

```python
categorical_columns = [
    "Participation in Extracurricular Activities",
    "Parent Education Level",
    "Passed"
]

for col in categorical_columns:
    print(f"\nColumn: {col}")
    print(df_clean[col].unique())
```
